```python
a, b = 0, 1

print("Fibonacci Series (10 terms):")

for i in range(10):
    print(a, end=" ")
    a, b = b, a + b
```

```python
num = int(input("Enter a number: "))

factorial = 1

if num < 0:
    print("Factorial is not defined for negative numbers.")
elif num == 0:
    print("Factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print("Factorial of", num, "is", factorial)
```

```python
rows = int(input("Enter number of rows: "))

for i in range(1, rows + 1):        # Outer loop: controls rows
    for j in range(1, i + 1):       # Inner loop: controls columns
        print("*", end=" ")         # Print * without newline
    print()
```

```python
# Simple Calculator using Functions

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error! Division by zero."
    else:
        return a / b


# Take input from user
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

print("\nSelect Operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

choice = input("Enter choice (1/2/3/4): ")

if choice == '1':
    print("Result:", add(num1, num2))
elif choice == '2':
    print("Result:", subtract(num1, num2))
elif choice == '3':
    print("Result:", multiply(num1, num2))
elif choice == '4':
    print("Result:", divide(num1, num2))
else:
    print("Invalid choice!")
```

```python
# Remove the first duplicate element from a list

my_list = [1, 2, 3, 2, 4, 3, 5]

seen = []       # Tracks elements we've already seen
result = []     # Final list without first duplicate

for item in my_list:
    if item not in seen:
        seen.append(item)    # First time seeing it — keep it
        result.append(item)
    else:
        seen.append(item)    # Already seen — skip ONLY the first duplicate
        my_list.remove(item) # Remove first occurrence of duplicate
        break                # Stop after removing first duplicate

print("Original List:", [1, 2, 3, 2, 4, 3, 5])
print("After Removing First Duplicate:", my_list)
```

```python
n = int(input("How many multiples of 3 do you want? "))

multiples = []

for i in range(1, n + 1):
    multiples.append(i * 3)

print("Multiples of 3:", multiples)
```
